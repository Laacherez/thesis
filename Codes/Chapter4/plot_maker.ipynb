```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Pe_range = np.linspace(0.1, 40, 10000)
kbT = 1
eta = 1.0
a = 1
D0 = kbT / 6 * np.pi * eta * a

Deff = D0 * (1 + 1 / 48 * Pe_range**2)

print(Deff)
```

```python
import matplotlib.ticker as mticker

fig, ax = plt.subplots(figsize=(4.5, 3))

ax.plot(Pe_range**2, Deff / D0, label="Eq.~4.9")

ax.set_xscale("log")
ax.set_yscale("log")

ax.set_xlabel(r"$\mathrm{Pe}^2$")
ax.set_ylabel(r"$D_{x,\mathrm{TA}}/D_0$")

ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f"{y:g}"))

ax.yaxis.set_minor_formatter(mticker.NullFormatter())

fig.tight_layout()

plt.legend(frameon=False)
plt.xlim(1e-2, 1e1)
plt.ylim(0.995, 1.1)
fig.savefig(
    "/Users/juls/Desktop/thesis/figures/chapter4/DeffD0.pdf",
    transparent=True,
)
plt.show()
```
